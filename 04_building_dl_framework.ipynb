# 构建深度学习框架

## 自动微分

```python
import numpy as np
```

```python
class Variable:
    """
        用于存放变量的类
    """
    def __init__(self, data):
        self.data = data
```

```python
x = Variable(data = np.array(1.0))
print(x.data)

x.data = np.array(2.0)
print(x.data)
```

```python
class Function:
    '''
    Function
    '''
    def __call__(self, value:Variable)->Variable:
        input_data = value.data
        output_data = self.forward(input_data)
        output = Variable(output_data)
        return output

    def forward(self, value):
        '''
        forward
        '''
        return NotImplementedError()
```

```python
class Square(Function):
    '''
    Square
    '''
    def forward(self, value):
        return value ** 2

x = Variable(np.array(10))
fn = Square()
y = fn(x)

print(type(y))
print(y.data)
```

```python
class Exp(Function):
    '''
    Exp
    '''
    def forward(self, value):
        return np.exp(value)
```

```python
fn_a = Square()
fn_b = Exp()
fn_c = Square()

var_x = Variable(np.array(0.5))
var_a = fn_a(var_x)
var_b = fn_b(var_a)
var_c = fn_c(var_b)

print(var_c.data)
```

```python
def numerical_diff(fn, x, eps = 1e-4) -> float:
    '''
    Numerical diff
    '''
    x_0 = Variable(x.data - eps)
    y_0 = fn(x_0)

    x_1 = Variable(x.data + eps)
    y_1 = fn(x_1)

    return (y_1.data - y_0.data) / (2 * eps)
```

```python
fn_0 = Square()
x_0 = Variable(np.array(2.0))
dy_0 = numerical_diff(fn_0, x_0)
print(dy_0)

def fn_1(x):
    '''
    (e^(x^2))^2
    '''
    fn_a = Square()
    fn_b = Exp()
    fn_c = Square()
    return fn_c(fn_b(fn_a(x)))

x_1 = Variable(0.5)
dy_1 = numerical_diff(fn_1, x_1)
print(dy_1)
```
